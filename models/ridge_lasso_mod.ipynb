```python
from load_datasets import load_train_test_sets, split_train_evaluate, create_submission_file
import pandas as pd
```

```python
X_train, X_test, train_target = load_train_test_sets()
```

```python
X_train.head()
```

### Training the model using Ridge

```python
from sklearn.linear_model import Ridge
```

```python
ridge_model = Ridge(alpha=10)
```

```python
ridge_mod, train_loss, val_loss = split_train_evaluate(ridge_model, X_train, train_target)
```

```python
ridge_preds = ridge_mod.predict(X_test)
ridge_preds
```

```python
create_submission_file(ridge_preds, 'ridge_model_sub')
```

### Train model using Lasso

```python
from sklearn.linear_model import Lasso
```

```python
lasso_model = Lasso(alpha=40)
```

```python
lasso_mod, train_loss, val_loss = split_train_evaluate(lasso_model, X_train, train_target)
```

```python
lasso_preds = lasso_mod.predict(X_test)
lasso_preds
```

```python
create_submission_file(lasso_preds, 'lasso_mod_sub')
```

### Train model using Elastic Net 

```python
from sklearn.linear_model import ElasticNet
```

```python
en_model = ElasticNet(alpha=0.01, l1_ratio=0.1)
```

```python
en_mod, train_loss, val_loss = split_train_evaluate(en_model, X_train, train_target)
```

```python
en_preds = en_mod.predict(X_test)
en_preds
```

```python
create_submission_file(en_preds, 'elastic_net_mod_sub')
```
